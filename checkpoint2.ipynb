```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stat
```

```python
data = pd.read_csv("./nominal_expenditures.csv")
df = pd.DataFrame(data)
df.head()
```

```python
plt.plot(df["Year"], df["Food sales at grocery stores with taxes and tips (millions of nominal U.S. dollars)"], marker = "o")
plt.plot(df["Year"], df["Food sales at full-service restaurants with taxes and tips (millions of nominal U.S. dollars)"], marker = "*", color = "red")
plt.xlabel = "Year"
plt.ylabel = "Food sales at grocery stores with taxes and tips (millions of nominal U.S. dollars)"
plt.title("Food sales at grocerry store vs Year")

plt.legend()
plt.show()
```

```python
plt.plot(df["Year"], df["Alcohol sales at eating and drinking places with taxes and tips (millions of nominal U.S. dollars)"], marker = "o")
plt.plot(df["Year"], df["Total alcohol-at-home sales with taxes and tips (millions of nominal U.S. dollars)"], marker = "*", color = 'red')
plt.plot(df["Year"], df["Food furnished and donated (millions of nominal U.S. dollars)"], marker = "*", color = 'green')

plt.xlabel = "Year"
plt.ylabel = "Food sales at grocery stores with taxes and tips (millions of nominal U.S. dollars)"
plt.title("Alcohol sales at eating and drinking places vs Year")

plt.legend()
plt.show()
```

```python
before = df[df["Year"] < 2020][
    "Total food-away-from-home expenditures with taxes and tips (millions of nominal U.S. dollars)"
]

after = df[df["Year"] >= 2020][
    "Total food-away-from-home expenditures with taxes and tips (millions of nominal U.S. dollars)"
]

t_stat, p_value = stat.ttest_ind(before, after, equal_var=False)

before_avg = before.mean()
after_avg = after.mean()

print("1997 - 2019 food-away-from-home expenditures : ", round(before_avg,2))
print("2020 - 2025 food-away-from-home expenditures : ", round(after_avg,2))

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Restaurant spending changed significantly.")
else:
    print("There is no significant difference in spending.")

x = ["1997 - 2019" ,"2020 - 2025" ]
y = [before_avg, after_avg]

plt.bar(x, y, color = 'green')

plt.xlabel = "Year"
plt.ylabel = "Total food-away-from-home expenditures with taxes and tips (millions of nominal U.S. dollars)"
plt.title("Total food-away-from-home average expenditures")


plt.show()
```
