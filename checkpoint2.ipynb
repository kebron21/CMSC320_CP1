```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stat
```

```python
data = pd.read_csv("./nominal_expenditures.csv")
df = pd.DataFrame(data)
df.head()
```

```python
df.info()

print("Missing values:", df.isna().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

df.describe()
```

```python
df.shape
```

```python
food_away = "Total food-away-from-home expenditures with taxes and tips (millions of nominal U.S. dollars)"
before = df[df["Year"] < 2020][food_away]

after = df[df["Year"] >= 2020][food_away]

t_stat, p_value = stat.ttest_ind(before, after, equal_var=False)

before_avg = before.mean()
after_avg = after.mean()

print("1997 - 2019 food-away-from-home expenditures : ", round(before_avg/1000,2), " Billion Dollars")
print("2020 - 2025 food-away-from-home expenditures : ", round(after_avg/1000,2) , " Billion Dollars" )

print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Restaurant spending changed significantly.")
else:
    print("There is no significant difference in spending.")

x = ["1997 - 2019" ,"2020 - 2025" ]
y = [before_avg, after_avg]

plt.bar(x, y)

plt.xlabel("Time Period")
plt.ylabel("Average Expenditures (Millions of Dollars)")
plt.title("Average Food-Away-From-Home Expenditures")




plt.show()
```

From 1997 to 2019, spending was $624.15 billion, while from 2020 to 2025, it was $1,193.09 billion. This means spending was much higher in the second period. The p-value is 0.000159, which is  smaller than 0.05. This tells us that the difference is statistically significant, meaning it is unlikely to be due to random chance alone. The negative t-statistic (-6.31) indicates the direction of the difference based on how the two groups were compared.Restaurant spending increased significantly in the more recent period compared with the earlier period.

```python
home = "Total food-at-home expenditures with taxes and tips (millions of nominal U.S. dollars)"
away = "Total food-away-from-home expenditures with taxes and tips (millions of nominal U.S. dollars)"


x = df[home] / 1000
y = df[away] / 1000


r, p = stat.pearsonr(x, y)
print("r = ",round(r,3))
print("P-value   = ", p)

slope, intercept = np.polyfit(x, y, 1)

fig, ax = plt.subplots(figsize=(7, 5))

ax.scatter(x, y, color="steelblue", edgecolor="black", s=60)
ax.plot(x, slope * x + intercept, color="darkred", linewidth=2, label=f"r = {r:.3f}")

ax.set_xlabel("Food-at-home spending (Billions of Dollars)")
ax.set_ylabel("Food away from home spending (Billions of Dollars)")
ax.set_title("Food at Home vs. Food Away From Home Spending")
ax.legend()
ax.grid(alpha=0.3)
plt.show()

```

The correlation between the two types of spending is 0.985, which indicates a very strong positive relationship. Both types of spending are measured in nominal dollars. As prices rise, people generally spend more money on goods and services, even if their actual consumption does not increase by the same amount. Therefore, some of the strong correlation may be explained by inflation and the shared upward trend in both spending categories, rather than one type of spending directly causing the other to increase. Although the correlation shows that the two variables move very closely together, it does not necessarily mean that changes in one category are responsible for changes in the other.

```python
grocery = "Food sales at grocery stores with taxes and tips (millions of nominal U.S. dollars)"

warehouse = "Food sales at warehouse clubs and supercenters with taxes and tips (millions of nominal U.S. dollars)"

groceryGrowth = df[grocery].pct_change() * 100
warehouseGrowth = df[warehouse].pct_change() * 100

warehouseGrowth = warehouseGrowth.iloc[1:]
groceryGrowth = groceryGrowth.iloc[1:]

print("Average Grocery Growth:", groceryGrowth.mean())
print("Average Warehouse Growth:", warehouseGrowth.mean())

t_stat, p_value = stat.ttest_rel(warehouseGrowth, groceryGrowth)

print("T-statistic:", t_stat)
print("P-value:", p_value)

plt.boxplot(
    [warehouseGrowth, groceryGrowth],
    tick_labels=["Warehouse Clubs", "Grocery Stores"]
)

plt.ylabel("Annual Percentage Growth")
plt.title("Annual Spending Growth by Store Type")
plt.show()


```

we compared annual spending growth at grocery stores and warehouse clubs and supercenters from 1998 to 2025. The results indicated  that warehouse clubs and supercenters had an average annual growth rate of 7.95%, compared to 3.03% for grocery stores. A paired t-test was performed to determine whether this difference was statistically significant. The test produced a p-value of 0.0000523, which is much less than the significance level of 0.05.Based on the result we reject the null hypothesis and conclude that warehouse clubs and supercenters experienced significantly higher average annual spending growth than grocery stores during the period studied. This may suggest that warehouse clubs and supercenters have become an increasingly important part of where Americans purchase food
